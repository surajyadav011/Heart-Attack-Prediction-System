# IQR Method(inter quartile method)

```python
# two method to delete outlier i) iqr II) z score method
```

```python
HR1 = HR.copy() # it makes copy 
```

```python
for col in HR1.columns:
    Q1 = HR1[col].quantile(0.25) # quantile 1 = 25 percent , Q1 = df['column_name'].quantile(0.25) to find quantile 1 
    Q3 = HR1[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_fence = Q3 + (IQR*1.5) # upper fence in boxplot
    lower_fence = Q1 - (IQR*1.5)
    HR1 = HR1[HR1[col].between(lower_fence,upper_fence)]  # remove outlier and keep value b/w lower and upper fence
```

# Z score method 

```python
HR2 = HR.copy()
```

```python
for col in HR2.columns:              # loop through columns
    STD = StandardScaler()           # create scaler
    HR2[col+'STD'] = STD.fit_transform(HR2[[col]])  # compute z-score
    HR2 = HR2[HR2[col+'STD'].between(-3,3)]  # keep non-outliers
    HR2.drop(col+'STD',axis=1,inplace=True) # remove temp column
```

```python
# x = patient health features
# y = heart attack result
x=HR2.drop('target',axis=1)
y=HR2.target
```

```python
std = StandardScaler()
norm = MinMaxScaler()
x_std = std.fit_transform(x) # standard  feature scaling
x_norm = norm.fit_transform(x) # normalization scaling
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x_norm, y, test_size=0.25)             # split data into x,y train and x,y test
# You can use x_std or x instead of x_norm; use whichever gives the best accuracy.
# 0.25 means 25% data → testing, 75% data → training
```

# Building  a Model

```python
LR.fit(xtrain,ytrain)
```
